```python
import pandas as pd
import matplotlib.pyplot as plt

my_data = pd.read_csv('Premier_League.csv')
my_data.head()
```

```python
# here we want to check if more possession (X) leads to more goals (Y)
plt.scatter(my_data['home_possessions'], my_data['Goals Home'])

# adding some basic labels
plt.title('Possession vs Goals (Home Teams)')
plt.xlabel('Possession %')
plt.ylabel('Goals Scored')

# display the plot
plt.show()
```

```python
# Using .mean() to get the average shots
avg_home_shots = my_data['home_shots'].mean()
avg_away_shots = my_data['away_shots'].mean()

# Putting data in simple lists
categories = ['Home Shots', 'Away Shots']
averages = [avg_home_shots, avg_away_shots]

# Plotting the bars
plt.bar(categories, averages)
plt.title('Average Shots: Home vs Away')
plt.ylabel('Average Number of Shots')

plt.show()
```

```python
# The attendance has commas like "25,000" so we must remove them first
my_data['attendance'] = my_data['attendance'].str.replace(',', '').astype(float)

# Drawing the histogram to see the distribution
plt.hist(my_data['attendance'])

plt.title('Attendance Distribution')
plt.xlabel('Number of Fans')
plt.ylabel('Number of Matches')

plt.show()
```
