```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('Premier_League.csv')

# I learned that Seaborn needs numbers, not strings with commas
df['attendance'] = df['attendance'].str.replace(',', '').astype(float)

sns.set_theme(style="whitegrid")
```

```python
# See the distribution of Home Possession
# kde=True adds a smooth line over the bars
sns.histplot(df['home_possessions'], kde=True, color='blue')

plt.title('Distribution of Possession for Home Teams')
plt.xlabel('Possession %')
plt.show()
```

```python
# Jointplot to see the relation between Shots on Target and Goals
# This shows a scatter plot and a histogram at the same time!
sns.jointplot(x='home_on', y='Goals Home', data=df, kind='reg', color='green')

plt.show()
```

```python
# Boxplot for Attendance
# The dots outside the "whiskers" are the outliers (huge stadiums)
sns.boxplot(x=df['attendance'], color='orange')

plt.title('Checking for Outliers in Attendance')
plt.xlabel('Number of Fans')
plt.show()
```

```python
# Correlation Heatmap
# I will choose only some columns to make it clear
columns_to_study = ['Goals Home', 'home_possessions', 'home_shots', 'home_on', 'home_fouls']
correlation_matrix = df[columns_to_study].corr()

# Plotting the heatmap
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm')

plt.title('What affects Home Goals the most?')
plt.show()
```
